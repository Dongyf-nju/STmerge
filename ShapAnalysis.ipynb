```python
from scipy.stats import gaussian_kde
from scipy.stats import pearsonr 
import statsmodels.api as sm

def add_nonlinear_trend(ax, x, y):
    # 计算LOWESS平滑曲线
    lowess = sm.nonparametric.lowess(y, x, frac=0.3, it=3)
    ax.plot(lowess[:,0], lowess[:,1], color='red', 
            linewidth=2, label='LOWESS')
    return ax 
def plot_density_scatter_matplotlib(ax, xlable, Xvar, Yvar, xlim, colormap):
    x = Xvar
    y = Yvar
    
    # 计算核密度估计
    xy = np.vstack([x, y])
    z = gaussian_kde(xy)(xy)
    # print(np.min(z) ,np.max(z))
    # 根据z值进行排序（颜色渐变效果）
    idx = z.argsort()
    x, y, z = x[idx], y[idx], z[idx]
    
    scatters= ax.scatter(x, y, c=z, cmap=colormap, s=5, edgecolor=None)
    # ax.set_title(f"{Xvar} vs {Yvar} Density Scatter Plot")
    ax.set_xlabel(f"{xlable} ({get_feature_unit(feature_name)})",fontsize=20)
    # 设置刻度标签大小
    ax.tick_params(axis='both', labelsize=20)
    # 设置x轴范围
    ax.set_xlim(xlim)
    # 添加Pearson相关系数
    corr_coef, p_value = pearsonr(Xvar, Yvar)
    ax.text(0.3, 0.97, f"ρ = {corr_coef:.2f}",#"\np = {p_value:.1e}", 
            transform=ax.transAxes, fontsize=18, va='top',
            bbox=dict(facecolor='none', edgecolor='none', alpha=0.8))
    # 展示非线性趋势
    # ax = add_nonlinear_trend(ax, x, y)
    # 调整刻度线参数：
    ax.minorticks_on()
    # 显示副刻度线
    ax.tick_params(axis="both", which="major", direction="in", width=0.9, length=5, color='#1C1C1C')
    # ax.tick_params(axis="both", which="minor", direction="in", width=0.55, length=1, color='#363636')
    
    # grid 设置网格线性
    ax.grid(True, which="major", linestyle="--", color="#1C1C1C", linewidth=0.75, alpha = 0.3)
    # ax.grid(True, which="minor", linestyle=":", color="gray", linewidth=0.55, alpha = 0.1)
    return scatters

# ------------------------------------可视化------------------------------------
FS_features = pd.read_csv(f'feature_importance_{MODEL_NAME}.csv')["Feature"].tolist()
# 提取前14个特征
FS_features = FS_features[0:MAX_DISPLAY]
FS_features_Newname = FS_features
num_cols = 5
num_rows = int(np.ceil(len(FS_features_Newname) / num_cols))
fig, axs = plt.subplots(nrows=num_rows, ncols=num_cols, figsize=(5*num_cols, 5*num_rows),dpi=300)
sub_title = 13
total_titlesize = 8
xy_labelsize = 7
cbar_titlesize = 8
cbar_lablesize = 8
title_size = 16
num_feature =MAX_DISPLAY
# 定义标签字符，生成 "a" 到 "t" 的字符
labels = [chr(i) for i in range(ord('a'), ord('a') + num_feature)]
import cmaps
colormap='viridis'
colormap= cmaps.NCV_jet
colormap= cmaps.percent_11lev
# 
idx=0
for feature_name in FS_features_Newname[0:num_feature]:
    if feature_name in features_NewName:
        
        row = idx // num_cols
        col = idx % num_cols
        if idx>num_feature:
            axs[row, col].axis('off')
        else:
            ax = axs[row, col] #plt.subplot(num_rows, num_cols, idx+1)  # 获取当前子图的轴
            TITLE = feature_name
            Yvar=shap_df[feature_name].values
            xlable = feature_name
            xlim = get_feature_xlim(feature_name)

            if feature_name in [ 'pH', 'TN', 'TP', 'SOC',  'CEC', 'TK']:
                Xvar=X_test_NewName[feature_name].values*0.01
                
            elif feature_name in ['BD']:
                Xvar=X_test_NewName[feature_name].values*0.001  
            elif feature_name in ['NLW', 'NSW', 'LW-Down', 'SW-Down']:
                Xvar=X_test_NewName[feature_name].values/24
            elif feature_name in ['PSFC']:
                Xvar=X_test_NewName[feature_name].values*0.01 # Pa -> hPa
                
            else:
                Xvar=X_test_NewName[feature_name].values
                

            scatters = plot_density_scatter_matplotlib(ax, xlable, Xvar, Yvar, xlim, colormap)
            # shap.dependence_plot(feature_name, shap_df[features].values, X_test, 
            #                       interaction_index=None, ax=axs[row, col], 
            #                       title=TITLE[0], alpha=0.2, label_size=22,axis_color='black',color="#4683B5")

            ax.axhline(y=0, color='red', linewidth=1., linestyle='--')

            ax.text(
                0.05, 0.97, 
                f"({labels[idx]})", transform=ax.transAxes, 
                fontsize=22, 
                # fontweight='bold',
                va='top', ha='left')
        idx=idx+1
        
# 隐藏剩余的子图
for i in range(idx, num_rows * num_cols):
    row = i // num_cols
    col = i % num_cols
    axs[row, col].axis('off')
# =============================================================================
# # ----------------------------绘制colorbar----------------------------
# 指定colorbar的位置和大小
cax = fig.add_axes([0.3, 0.05, 0.4, 0.01])
# 第一个参数：子图左下角的x坐标，表示子图距离整个图形左边界的距离占整个图形宽度的比例。
# 第二个参数：子图左下角的y坐标，表示子图距福整个图形底边界的距离占整个图形高度的比例。
# 第三个参数：子图的宽度，表示子图宽度占整个图形宽度的比例。
# 第四个参数：子图的高度，表示子图高度占整个图形高度的比例。
cbar = fig.colorbar(scatters, shrink=0.6,fraction=0.04, pad=0, orientation='horizontal', ax=axs,cax=cax)#, ticks=levels)
# 设置坐标轴标签字体大小
cbar.ax.tick_params(labelsize=18)
plt.subplots_adjust(wspace=0.3,hspace=0.3)
save_name = os.path.join(Figure_path, f"CN_{MODEL_NAME}_dependence_plot_subplot.png")
# plt.savefig(save_name, dpi=600, bbox_inches='tight', pad_inches=0.1)
plt.show()
```
